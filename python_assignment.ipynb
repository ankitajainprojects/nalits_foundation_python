```python
start = 10
end = 30

def primes_in_interval(start, end):
    primes = []
    for num in range(start, end+1):
        if is_prime(num):
            primes.append(num)
    return primes

def is_prime(num):
    is_prime = True

    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break

    return is_prime    

print("Primes:", primes_in_interval(start, end))


            
```

```python
text = "Hello World"
vowels = 0
consonants = 0
vowels_string = "aeiou"

for char in text:
    if char.isalpha():
        if vowels_string.find(char.lower()) != -1:
            vowels = vowels + 1
        else:
            consonants = consonants + 1    



print(f"Vowels: {vowels}, Consonants: {consonants}")
```
